# ฟังเสียง TTS อารมณ์ (v4 + ระบบผสม)
ต้องรัน `./TTS-Lab/run_tts_native.sh` ค้างไว้ก่อน (server ที่ localhost:8001)
รอบแรกของแต่ละเสียงอาจใช้ ~20 วินาทีบน CPU

```python
import requests, io
from IPython.display import Audio, display, HTML
URL = "http://localhost:8001"
print(requests.get(URL + "/v1/health").json())

def speak(text, voice="female", emotion="neutral", intensity=0.6, speed=1.0):
    r = requests.post(URL + "/v1/speak", json=dict(text=text, voice=voice, emotion=emotion,
                      intensity=intensity, speed=speed), timeout=300)
    r.raise_for_status()
    return r.content, r.headers.get("x-engine")
```

## 1) ประโยคเดียว เทียบทุกอารมณ์

```python
TEXT   = "วันนี้ฉันดีใจมาก"
VOICE  = "female"      # male / female
EMOS   = ["neutral", "happy", "sad", "angry", "fear", "disgust", "surprise"]
INTENS = 0.6

for e in EMOS:
    wav, eng = speak(TEXT, VOICE, e, INTENS)
    display(HTML(f"<b>{e}</b> <small>({eng})</small>"))
    display(Audio(wav, autoplay=False))
```

## 2) หลายประโยค × หลายอารมณ์ (ตารางฟัง)

```python
TESTS = [
    ("วันนี้ฉันดีใจมากที่สอบผ่าน", "happy"),
    ("ฉันเสียใจมากที่เขาจากไป", "sad"),
    ("ทำไมทำแบบนี้ ฉันโมโหมากแล้วนะ", "angry"),
    ("วันนี้อากาศดี", "neutral"),
]
VOICES = ["female", "male"]

for text, emo in TESTS:
    for v in VOICES:
        wav, eng = speak(text, v, emo)
        display(HTML(f"<b>{emo}</b> · {v} · {text}"))
        display(Audio(wav))
```

## 3) ลองเองอิสระ

```python
text = "ขอบคุณมากครับ"
voice = "male"
emotion = "happy"    # neutral/happy/sad/angry/fear/disgust/surprise
intensity = 0.6
speed = 1.0

wav, eng = speak(text, voice, emotion, intensity, speed)
print(eng)
Audio(wav)
```
